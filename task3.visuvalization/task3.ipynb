```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
df=pd.read_csv("employees.csv")
print(df)
```

```python
print("Dataset shape:", df.shape)
print(df.head(5))
```

```python
department_salary=df.groupby("department")["salary"].mean()
print(department_salary)
```

```python
department_salary.plot(kind="bar")
plt.title("Average salary by Department")
plt.xlabel("department")
plt.ylabel("average salary($)")
plt.show()
```

```python
department_salary.plot(kind="bar",color="yellow",label="average salary")
plt.title("Average salary by department")
plt.xlabel("department")
plt.ylabel("average salary($)")
plt.legend()
plt.tight_layout()
plt.savefig("reports/figures/01_department_salary.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(
    df["age"],bins=5,color="yellow",edgecolor="black",label="Employees"
)
plt.title("Employe age distribution")
plt.xlabel("Age(years)")

plt.ylabel("Numer of employes(count)")
plt.legend()
plt.tight_layout()
plt.savefig("reports/figures/02_age_histogram.png")
plt.show()

```

```python
plt.figure(figsize=(8,5))
plt.scatter(
    df["age"],
    df["salary"],
    color="blue",
    label="Employees"
)
plt.title("Relationship between salary and age")
plt.xlabel("age(years)")
plt.ylabel("salary($)")
plt.legend()
plt.tight_layout()
plt.savefig("reports/figures/03_age_salary_scatter.png")
plt.show()

```

```python
df_sorted = df.sort_values("age")

print(df_sorted[["name", "age", "salary"]])
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df_sorted["age"],
    df_sorted["salary"],
    color="skyblue",
    marker="o",
    label="Employee Salary"
)

plt.title("Salary Trend by Employee Age")
plt.xlabel("Age (Years)")
plt.ylabel("Salary (₹)")

plt.legend()
plt.tight_layout()

plt.savefig("reports/figures/04_salary_line_chart.png")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

df.boxplot(
    column="salary",
    by="department",
    color="red"
)

plt.title("Salary Distribution by Department")
plt.suptitle("")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")

plt.tight_layout()

plt.savefig("reports/figures/05_salary_boxplot.png")

plt.show()
```
