```python
print(df[["age","salary"]].agg(["mean", "median", "min", "max", "std"]))
```

```python
department_analysis=df.groupby("department").agg(
    employee_count=("employee_id", "count"),
    average_age=("age", "mean"),
    average_salary=("salary", "mean"),
    maximum_salary=("salary", "max"),
    minimum_salary=("salary", "min")
)
print(department_analysis)
```

```python
numeric_df = df[["age", "salary", "experience_years"]]

correlation = numeric_df.corr()

print(correlation)
```

```python

print("\nBUSINESS QUESTIONS AND ANSWERS")

total_employees = df["employee_id"].count()
print("1. Total Employees:", total_employees)

average_salary = df["salary"].mean()
print("2. Average Salary:", average_salary)

highest_salary = df["salary"].max()
highest_paid_employee = df.loc[df["salary"].idxmax(), "name"]
print("3. Highest Paid Employee:", highest_paid_employee)
print("   Highest Salary:", highest_salary)

department_salary = df.groupby("department")["salary"].mean()
highest_department = department_salary.idxmax()
highest_department_salary = department_salary.max()

print("4. Highest Paying Department:", highest_department)
print("   Department Average Salary:", highest_department_salary)

average_experience = df["experience_years"].mean()
print("5. Average Experience:", average_experience)
```

```python
department_budget = pd.DataFrame({
    "department": ["IT", "HR", "Sales", "Finance"],
    "department_budget": [500000, 300000, 250000, 400000]
})

print("\nDepartment Budget Table")
print(department_budget)
merged_df = pd.merge(
    df,
    department_budget,
    on="department",
    how="left"
)

print("\nMerged Employee and Budget Data")
print(merged_df[["name", "department", "salary", "department_budget"]].head())
```
